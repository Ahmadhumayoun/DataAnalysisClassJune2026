```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
Xaxis = np.array([1,2,3,4,5,6,7,8])
Yaxis = np.array([10,20,30,40,50,60,70,80])
Ayaxis = np.array([15,85,35,45,55,65,75,85])
```

```python
plt.xlabel("Independant Values")
plt.ylabel("Dependant Values")
plt.title("Demo Of Matplotlib")
plt.grid()
plt.legend()
plt.plot(Xaxis,Yaxis,
         marker="o",
         markersize = 15,
         markerfacecolor="red",
         linestyle = "--",
        color = "purple",
        linewidth = 4,
        label="Team A")

plt.plot(Xaxis,Ayaxis,
         marker="*",
         markersize = 15,
         markerfacecolor="blue",
         linestyle = "--",
        color = "red",
        linewidth = 4,
        label="Team B")
```

```python
c = ["A","B","C","D"]
va = [10,24,18,25]

plt.bar(c,va,width=.2,
        color =['#2b7bba', '#f28e2b', '#2ca02c', '#d62728'],
        edgecolor="blue",
        )
plt.show()
```
