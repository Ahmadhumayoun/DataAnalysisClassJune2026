```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
Xaxis = np.array([1,2,3,4,5,6,7,8])
Yaxis = np.array([10,20,30,40,50,60,70,80])
Ayaxis = np.array([15,85,35,45,55,65,75,85])
```

```python
plt.xlabel("Independant Values")
plt.ylabel("Dependant Values")
plt.title("Demo Of Matplotlib")
plt.grid()
plt.legend()
plt.plot(Xaxis,Yaxis,
         marker="o",
         markersize = 15,
         markerfacecolor="red",
         linestyle = "--",
        color = "purple",
        linewidth = 4,
        label="Team A")

plt.plot(Xaxis,Ayaxis,
         marker="*",
         markersize = 15,
         markerfacecolor="blue",
         linestyle = "--",
        color = "red",
        linewidth = 4,
        label="Team B")
```

```python
c = ["A","B","C","D"]
va = [10,24,18,25]

plt.bar(c,va,width=.2,
        color =['#2b7bba', '#f28e2b', '#2ca02c', '#d62728'],
        edgecolor="blue",
        )
plt.show()
```

```python
li = [35,15,25,25]
cat = ["Apple","Date","Banana","Peach"]
cusColors= ['red', 'blue', 'purple', 'blue']
exPer = [.1,0,.3,0]
plt.pie(li,
        labels=cat,
        colors=cusColors,
        explode=exPer,
        autopct="%1.1f%%",
        textprops={'fontsize': 12, 'weight': 'bold'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 1})
```

```python

fig , axs = plt.subplots(4,5)
li = [35,15,25,25]
cat = ["Apple","Date","Banana","Peach"]
cusColors= ['red', 'blue', 'purple', 'blue']
exPer = [.1,0,.3,0]
axs[0,2].pie(li,
        labels=cat,
        colors=cusColors,
        explode=exPer,
        autopct="%1.1f%%",
        textprops={'fontsize': 12, 'weight': 'bold'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 1})

c = ["A","B","C","D"]
va = [10,24,18,25]

axs[0,1].bar(c,va,width=.2,
        color =['#2b7bba', '#f28e2b', '#2ca02c', '#d62728'],
        edgecolor="blue",
        )

plt.tight_layout
```
