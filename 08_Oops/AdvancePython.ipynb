```python
def calculator(operation, num1, num2):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "m":
        return num1 * num2
    elif operation == "divide":
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero is not allowed."
    else:
        return "Error: Invalid operation. Please choose from 'add', 'subtract', 'multiply', or 'divide'."
```

```python
calculator("m", 10, 5)
```

```python


dirtPrice = ["100,00","200,00"]
class SheetAnalysis:
    def cleanPrice(self,ListToclean,elementToRemove):
        cleanedList = []
        for price in ListToclean:
            cleanedPrice = price.replace(elementToRemove, "")
            cleanedList.append(cleanedPrice)
        return cleanedList
```

```python
sa = SheetAnalysis()
cleanedPrices = sa.cleanPrice(dirtPrice, ",")
print(cleanedPrices)
```

```python
class bankAccount:
    def __init__(self, balance):
        self.balance = balance

    def checkBalance(self):
        return self.balance
    def deposit(self, amount):
        self.balance += amount
        return self.balance
        
```

```python
b = bankAccount(5000)
print(b.checkBalance())
b.deposit(2000)
print(b.checkBalance())
```

```python
class grandparentClass:

    def functionFromParent(self):
        print("This is a method from the parent class.")

class parentClass(grandparentClass):
    def functionFromChild(self):
        print("This is a method from the child class.")
class ChildClass(parentClass):
    pass


c = ChildClass()
c.functionFromChild()
c.functionFromParent()
```

```python
class parent1:
    def functionFromParent1(self):
        print("This is a method from the parent class 1.")

class parent2:
    def functionFromParent2(self):
        print("This is a method from the parent class 2.")

class hybridChildClass(parent1, parent2):
    pass

hyb = hybridChildClass()
hyb.functionFromParent1()
hyb.functionFromParent2()
```
