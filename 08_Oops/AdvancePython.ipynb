```python
def calculator(operation, num1, num2):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "m":
        return num1 * num2
    elif operation == "divide":
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero is not allowed."
    else:
        return "Error: Invalid operation. Please choose from 'add', 'subtract', 'multiply', or 'divide'."
```

```python
calculator("m", 10, 5)
```

```python
import pandas as pd

class Account:
    def __init__(self, accountNumber, fname, lname, password, address, studentClass, city, balance):
        self.accountNumber = accountNumber
        self.fname = fname
        self.lname = lname
        self.password = password
        self.address = address
        self.studentClass = studentClass
        self.city = city
        self.balance = balance

    def display(self):
        print("First Name :", self.fname)
        print("Last Name :", self.lname)
        print("Address :", self.address)
        print("Class :", self.studentClass)
        print("City :", self.city)
        print("Balance :", self.balance)

    def checkBalance(self):
        return self.balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return self.balance
        return None
```

```python
class Bank:
    def __init__(self, filename="../03_Pandas_Data_Analysis/bank.csv"):
        self.filename = filename
        self.accounts = {}
        self.load()

    def load(self):
        df = pd.read_csv(self.filename)
        for i in range(len(df)):
            account = str(df["AccountNumber"][i])
            self.accounts[account] = Account(
                account,
                df["FName"][i],
                df["LName"][i],
                str(df["Password"][i]),
                df["Address"][i],
                df["Class"][i],
                df["City"][i],
                float(df["Balance"][i])
            )

    def save(self):
        rows = []
        for account in self.accounts:
            acc = self.accounts[account]
            rows.append([acc.accountNumber, acc.fname, acc.lname, acc.password,
                         acc.address, acc.studentClass, acc.city, acc.balance])
        df = pd.DataFrame(rows, columns=["AccountNumber", "FName", "LName",
                                         "Password", "Address", "Class", "City", "Balance"])
        df.to_csv(self.filename, index=False)
        print("Data Saved")

    def createAccount(self):
        account = input("Account Number: ")
        if account in self.accounts:
            print("Account Already Exists")
        else:
            fname = input("First Name: ")
            lname = input("Last Name: ")
            password = input("Password: ")
            address = input("Address: ")
            studentClass = input("Class: ")
            city = input("City: ")
            balance = float(input("Opening Balance: "))
            self.accounts[account] = Account(account, fname, lname, password,
                                             address, studentClass, city, balance)
            print("Account Created")

    def login(self):
        account = input("Account Number: ")
        password = input("Password: ")
        if account in self.accounts and password == self.accounts[account].password:
            self.accountMenu(account)
        else:
            print("Invalid Account or Password")

    def accountMenu(self, account):
        while True:
            print("\n1. View Details")
            print("2. Check Balance")
            print("3. Deposit")
            print("4. Withdraw")
            print("5. Logout")
            option = input("Choice: ")
            acc = self.accounts[account]
            if option == "1":
                acc.display()
            elif option == "2":
                print("Balance :", acc.checkBalance())
            elif option == "3":
                amount = float(input("Amount: "))
                print("New Balance :", acc.deposit(amount))
            elif option == "4":
                amount = float(input("Amount: "))
                result = acc.withdraw(amount)
                if result is None:
                    print("Insufficient Balance")
                else:
                    print("Remaining Balance :", result)
            elif option == "5":
                break
            else:
                print("Invalid Choice")
```

```python
def main_menu(bank):
    while True:
        print("\n1. Create Account")
        print("2. Login")
        print("3. Exit")
        choice = input("Enter Choice: ")
        if choice == "1":
            bank.createAccount()
        elif choice == "2":
            bank.login()
        elif choice == "3":
            bank.save()
            break
        else:
            print("Invalid Choice")

bank = Bank()
main_menu(bank)
```
