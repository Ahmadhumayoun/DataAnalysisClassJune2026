# Matplotlib Fundamentals — Data Visualization

A chart can communicate in one glance what a table needs paragraphs for. **Matplotlib** is Python's classic plotting library — pandas uses it under the hood for its own plots.

**What you will learn:**

- Line plots with full styling (markers, colors, line styles, labels)
- Bar charts
- Pie charts with explode and percentages
- Multi-plot figures with `plt.subplots()`
- Common pitfalls: legend placement, `tight_layout()`, `plt.show()`

## Setup

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
# Generate our sample data with NumPy
Xaxis = np.array([1,2,3,4,5,6,7,8])
Yaxis = np.array([10,20,30,40,50,60,70,80])
Ayaxis = np.array([15,85,35,45,55,65,75,85])
```

## Line Plot — Two Teams Over Time

A **line plot** shows how values change across a sequence (time, index, position). Two series can share one axes — each with its own `label` so the legend can tell them apart.

> **Pitfall:** call `plt.legend()` *after* you've plotted the labelled series — otherwise matplotlib has no labels to show (you'd get a "No artists with labels found" warning).

```python
# Plot BOTH teams on the same figure
plt.plot(Xaxis, Yaxis,
         marker="o",
         markersize=15,
         markerfacecolor="red",
         linestyle="--",
         color="purple",
         linewidth=4,
         label="Team A")

plt.plot(Xaxis, Ayaxis,
         marker="*",
         markersize=15,
         markerfacecolor="blue",
         linestyle="--",
         color="red",
         linewidth=4,
         label="Team B")

# Axes labels, grid and legend (AFTER the labelled plots!)
plt.xlabel("Independent Values")
plt.ylabel("Dependent Values")
plt.title("Demo of Matplotlib")
plt.grid()
plt.legend()
plt.show()
```

## Bar Chart — Categories vs. Values

A **bar chart** compares categories. Each bar gets its own color from the list — and `edgecolor` outlines them:

```python
c = ["A","B","C","D"]
va = [10,24,18,25]

plt.bar(c, va, width=.2,
        color=['#2b7bba', '#f28e2b', '#2ca02c', '#d62728'],
        edgecolor="blue")
plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Bar Chart Demo")
plt.show()
```

## Pie Chart — Parts of a Whole

A **pie chart** shows proportions. `explode` pulls a slice out, `autopct` prints percentages on each slice:

```python
li = [35,15,25,25]
cat = ["Apple","Date","Banana","Peach"]
cusColors = ['red', 'blue', 'purple', 'blue']
exPer = [.1, 0, .3, 0]

plt.pie(li,
        labels=cat,
        colors=cusColors,
        explode=exPer,
        autopct="%1.1f%%",
        textprops={'fontsize': 12, 'weight': 'bold'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 1})
plt.title("Fruit Pie Chart Demo")
plt.show()
```

## Subplots — Many Charts in One Figure

`plt.subplots(rows, cols)` creates a **grid of axes**. Pick one with `axs[r, c]` and draw into it exactly like the top-level figure.

> **Pitfall:** `plt.tight_layout()` needs parentheses! The original code had `plt.tight_layout` — valid Python, but it just *references* the function instead of calling it, so nothing happens.

### A 4×5 Grid of Pie Charts

The classic trick for showing every column of a dataset: loop over the grid and fill each cell with the same pie chart.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axs = plt.subplots(4, 5)

li = [35, 15, 25, 25]
cat = ["Apple", "Date", "Banana", "Peach"]
cusColors = ['red', 'blue', 'purple', 'blue']
exPer = [.1, 0, .3, 0]

# Fill the whole 4x5 grid with pie charts
for row in range(4):
    for col in range(5):
        axs[row, col].pie(li,
                          labels=cat,
                          colors=cusColors,
                          explode=exPer,
                          autopct="%1.1f%%",
                          textprops={'fontsize': 8, 'weight': 'bold'},
                          wedgeprops={'edgecolor': 'black', 'linewidth': 1})

plt.suptitle("20 Mini Pie Charts")      # one title for the whole figure
plt.tight_layout()                      # fix spacing — note the parentheses!
plt.show()
```

### Mixing Chart Types in One Figure

Each axes in the grid is independent — put a bar chart in one cell and a pie chart in another:

```python
fig, axs = plt.subplots(1, 3, figsize=(12, 4))

c = ["A", "B", "C", "D"]
va = [10, 24, 18, 25]

axs[0].bar(c, va, color=['#2b7bba', '#f28e2b', '#2ca02c', '#d62728'], edgecolor="blue")
axs[0].set_title("Bar Chart")

axs[1].plot(Xaxis, Yaxis, marker="o", color="purple", label="Team A")
axs[1].plot(Xaxis, Ayaxis, marker="*", color="red", label="Team B")
axs[1].set_title("Line Chart")
axs[1].legend()

axs[2].pie(li, labels=cat, autopct="%1.1f%%",
           wedgeprops={'edgecolor': 'black', 'linewidth': 1})
axs[2].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

## 🎯 Key Takeaways

- `plt.plot(x, y, label=..., marker=..., linestyle=..., linewidth=...)` — full styling in one call.
- `label=` + `plt.legend()` after plotting → labelled legend.
- `plt.bar()` for categories, `plt.pie()` for proportions (`explode`, `autopct`).
- `plt.subplots(rows, cols)` gives a grid of axes (`axs[r, c]`) for multi-plot figures.
- `plt.tight_layout()` (with parentheses!) fixes cramped spacing.
- Always finish with `plt.show()` so the figure renders in the notebook.
