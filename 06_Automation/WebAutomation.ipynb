```python
from selenium import webdriver as wd
from selenium.webdriver.common.by import By 
import time as t


driver = wd.Chrome()
driver.get("https://duramolab.com/")
signInButtonFromHomePage = "/html/body/section[1]/div/div/a[2]"
emailField = "/html/body/div/[6]/div/form/div[1]/input"
passField = "/html/body/div[6]/div/form/div[2]/input"
SignInButtonInsideSigninPage= "/html/body/div[6]/div/form/button"

t.sleep(3)
driver.find_element(By.XPATH,signInButtonFromHomePage).click()
t.sleep(3)
driver.find_element(By.XPATH,emailField).send_keys("ahmadonfiver@gmail.com")
driver.find_element(By.XPATH,passField).send_keys("/////")
t.sleep(1)
driver.find_element(By.XPATH,SignInButtonInsideSigninPage).click()
t.sleep(30)
driver.quit
```
