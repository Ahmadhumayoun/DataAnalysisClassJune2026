## Conditional Statements (if / elif / else)

Conditionals allow your program to make decisions based on boolean expressions.
- `if` — checks a condition
- `elif` — checks another condition if the first is False
- `else` — runs when all conditions are False


````python
if number1 > number2:
    print("Number is greater than number2")
else:
    print("Number is not greater than number2")
````

````python
# Simple if-else: check which number is larger
#Talke Two numbers from user and take input from user which operation to perfrom
# Check if results is even or odd and print the result
n1 = int(input("Enter Number 1: "))
n2 = int(input("Enter Number 2: "))
operation = input("Enter operation (+, -, *, /): ")

if operation == "+":
    result = n1 + n2
    if result % 2 == 0:
        print(f"The result of {n1} + {n2} = {result} and it is even")
    else:
        print(f"The result of {n1} + {n2} = {result} and it is odd")
elif operation == "-":
    result = n1 - n2    
    if result % 2 == 0:
        print(f"The result of {n1} - {n2} = {result} and it is even")
    else:
        print(f"The result of {n1} - {n2} = {result} and it is odd")

elif operation == "*":
    result = n1 * n2
    if result % 2 == 0:
        print(f"The result of {n1} * {n2} = {result} and it is even")
    else:
        print(f"The result of {n1} * {n2} = {result} and it is odd")
elif operation == "/":
    result = n1 / n2
    if int(result) % 2 == 0:
        print(f"The result of {n1} / {n2} = {result} and it is even")
    else:
        print(f"The result of {n1} / {n2} = {result} and it is odd")
else:
    print("Invalid operation")
    result = None
````

````python
# Combining conditions with logical operators
age = 22
has_license = True
has_id = False

# Both conditions must be True (and)
if age >= 18 and has_license:
    print("You can drive!")
else:
    print("You cannot drive")

# At least one condition must be True (or)
if has_license or has_id:
    print("You have some form of identification")
else:
    print("You have no ID")
````

````python
# Grading system using nested conditionals
marks = 85

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks} → Grade: {grade}")
````

## List Comprehensions

````python
fruits = {"apple", "banana", "orange", "apple"}
print(fruits)            # {'banana', 'orange', 'apple'} — no duplicates!

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(set_a | set_b)     # {1, 2, 3, 4, 5, 6} — union
print(set_a & set_b)     # {3, 4} — intersection
print(set_a - set_b)     # {1, 2} — difference
````

````python
# Variable scope — local vs global
global_var = "I am global"   # module-level scope

def my_function():
    local_var = "I am local"  # only exists inside function
    print(f"Inside function: {global_var}")
    print(f"Inside function: {local_var}")

my_function()
print(f"Outside function: {global_var}")
# print(local_var)  # This would cause NameError!
````

````python
# *args — variable number of positional arguments
def sum_all(*args):
    """Sum any number of arguments."""
    total = 0
    for num in args:
        total += num
    return total

print(f"sum_all(1, 2, 3) = {sum_all(1, 2, 3)}")
print(f"sum_all(10, 20, 30, 40, 50) = {sum_all(10, 20, 30, 40, 50)}")

# **kwargs — variable number of keyword arguments
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

print("User Info:")
print_info(name="Ahmad", age=25, city="Dubai")
````

````python
# Recursion — a function that calls itself
# Compute factorial: n! = n * (n-1) * (n-2) * ... * 1

def factorial(n):
    # Base case: stop when n is 0 or 1
    if n <= 1:
        return 1
    # Recursive case: n * factorial of (n-1)
    return n * factorial(n - 1)

# Also demonstrate the iterative version
def factorial_iterative(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(f"Factorial of 5 (recursive): {factorial(5)}")
print(f"Factorial of 5 (iterative): {factorial_iterative(5)}")
````

````python
# Lambda — small anonymous functions
# Syntax: lambda arguments: expression

# Lambda with map — apply to each element
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(f"Doubled: {doubled}")

# Lambda with filter — keep elements matching condition
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Evens: {evens}")

# Lambda with sorted — custom sort key
names = ["Ahmad", "Ali", "Muhammad", "Sarah"]
sorted_by_length = sorted(names, key=lambda n: len(n))
print(f"Sorted by name length: {sorted_by_length}")
````
