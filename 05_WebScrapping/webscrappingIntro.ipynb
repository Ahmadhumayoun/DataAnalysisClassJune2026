```python
import requests as r
from bs4 import BeautifulSoup as bs


portfolioResponse = r.get("https://www.scrapethissite.com/pages/simple/")

soup = bs(portfolioResponse.content,"html.parser")
# print(soup.prettify())
with open("ScrapeThisSiteResponse.html","w")as file:
    file.write(str(soup))

```

```python
with open("ScrapeThisSiteResponse.html","r") as file:
    soup = bs(file.read(),"html.parser")
    print(soup.find("title").text)
    print(soup.find("p",class_="",id_="jkads").text)
```
