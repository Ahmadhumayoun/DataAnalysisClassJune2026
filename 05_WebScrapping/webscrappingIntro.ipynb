```python
import pandas as pd
with open("ScrapeThisSiteResponse.html","r") as file:
    soup = bs(file.read(),"html.parser")
    # print(soup.find("title").text)
    countryNames = soup.find_all("h3",class_="country-name")
    countryCapitals= soup.find_all("span",class_="country-capital")
    countryPopulation= soup.find_all("span",class_="country-population")
    countryArea = soup.find_all("span",class_="country-area")
    "============================//=============================="
    CountryInfo  = {"Country Name": [] , "Country Capital" : [] , "Country Population" : [] , "Country Area" : [] }
    for i in range(0,len(countryNames)):
        CountryInfo["Country Name"].append(countryNames[i].text.strip())
        CountryInfo["Country Capital"].append(countryCapitals[i].text.strip())
        CountryInfo["Country Population"].append(countryPopulation[i].text.strip())
        CountryInfo["Country Area"].append(countryArea[i].text.strip())

    countryInfoDataFram = pd.DataFrame(CountryInfo)
    countryInfoDataFram.to_excel("CountryData.xlsx",index=False)
```

```python
baseurl = "https://www.deckshop.pro"

re = r.get(f"{baseurl}/card/list")

soup = bs(re.content,"html.parser")

imagesUrl = soup.find_all("img",class_="card")

ImageUrlData = {"Name":[],"Url":[]}
for eachUrl in imagesUrl:
    # print(eachUrl)
    ImageUrlData["Name"].append(str(eachUrl.get("alt")).strip())
    ImageUrlData["Url"].append(str(eachUrl.get("src")).strip())

import pandas as pd

imagedf = pd.DataFrame(ImageUrlData)
```

```python
# Scrapping Image
from tqdm import tqdm
for i in tqdm(range(len(ImageUrlData["Url"])),desc = "Downloading Image ... : ",colour="green"):
    imgUrl = ImageUrlData["Url"][i]
    re = r.get(f"{baseurl}{imgUrl}")
    imageName = ImageUrlData["Name"][i]
    with open(f"ScrappedImages/{imageName}.png","wb") as f:
        f.write(re.content)
```
