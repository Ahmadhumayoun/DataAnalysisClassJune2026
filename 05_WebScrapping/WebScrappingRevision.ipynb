```python
import requests as req
from bs4 import BeautifulSoup as bs
baseUrl = "https://www.deckshop.pro/"
response = req.get(baseUrl)
soup = bs(response.content, "html.parser")
with open("deskShopOffline.html","w") as file:
    file.write(soup.prettify())
```

```python
with open("ScrapeThisSiteOffline.html","r") as file:
    content = file.read()
    # print(type(content))
    soup = bs(content, "html.parser")
    countryNames = soup.find_all("h3",class_="country-name")
    for eachh3 in countryNames:
        print(eachh3.text.strip())

    
```

```python
from tqdm import tqdm
with open("deskShopOffline.html","r") as file:
    content = file.read()
    baseUrl = "https://www.deckshop.pro"
    soup = bs(content, "html.parser")
    allImages = soup.find_all("img")
    for eachImage in tqdm(allImages,desc="Downloading Images"):
        eachImageUrl = eachImage.get("src")
        eachImageName = eachImage.get("alt")
        ImageContent = req.get(baseUrl+eachImageUrl).content
        with open(f"ScrappedImages/{eachImageName}.png", "wb") as file:
            file.write(ImageContent)
```
