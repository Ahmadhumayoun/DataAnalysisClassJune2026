# File Handling in Python

Python can read from and write to files on disk. The `open()` function is used to open a file and returns a file object. The `with` statement ensures the file is properly closed after use.

**File modes:**
- `"r"` — read (default)
- `"w"` — write (overwrites existing file)
- `"a"` — append (adds to end)
- `"r+"` — read + write
- `"b"` — binary mode (e.g., `"rb"`, `"wb"`)

**What you will learn:**

- Read and write text files with `open()` and the `with` statement
- Choose the right file mode (`r`, `w`, `a`, ...)
- Store structured data as JSON and CSV
- Handle file errors gracefully with `try/except`
- Work with paths using `os` and the modern `pathlib` module
- Build a working user database (login/signup) stored on disk

### Why File Handling Matters

Being able to **read from** and **write to** files is essential for real-world programming. You'll need it for:

- Saving program output and logs
- Loading configuration files
- Processing datasets (CSV, JSON, Excel)
- Reading/writing user databases
- Interacting with other applications

Python's `open()` function and the `with` statement make file I/O clean and safe.


### File Modes Explained

The second argument to `open()` specifies the **mode**:

| Mode | Description |
|------|-------------|
| `"r"`  | Read (default) — file must exist |
| `"w"`  | Write — **creates/overwrites** file |
| `"a"`  | Append — adds to the end of existing file |
| `"x"`  | Exclusive creation — fails if file exists |
| `"r+"` | Read + Write (no truncation) |
| `"w+"` | Read + Write (truncates file) |
| `"a+"` | Read + Append |
| `"b"`  | Binary mode (add to any mode: `"rb"`, `"wb"`) |

**Always** use the `with` statement — it automatically closes the file, even if an error occurs.



````python
msg="Ahmad here !"
with open("introduction.txt","w") as file:
    file.write(msg)
````

````python
# .read() reads the entire file content into a string
# 'r' mode opens the file for reading (default)
# Writing multiple lines at once
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
with open("multi_line.txt", "w") as f:
    f.writelines(lines)

# Verify by reading
with open("multi_line.txt", "r") as f:
    print("File contents:")
    print(f.read())
````

````python
# Demonstrate different file modes
# 'w' — write mode (creates new file or overwrites existing)
with open("mode_demo.txt", "w") as f:
    f.write("This was written in 'w' mode.\n")

# 'a' — append mode (adds to existing content)
with open("mode_demo.txt", "a") as f:
    f.write("This was appended in 'a' mode.\n")

# 'r' — read mode
with open("mode_demo.txt", "r") as f:
    content = f.read()
    print("Content:", content)
````

## Writing to Files

Use `open("filename", "w")` to write. **Warning:** `"w"` mode overwrites the file if it exists!

```python
with open("output.txt", "w") as file:
    file.write("Hello, World!\n")
    file.writelines(["line1\n", "line2\n"])
```

To append instead of overwrite, use `"a"` mode.


````python
with open("introduction.txt","r") as file:
    print(file.read())
````

````python
# 'r' mode — read existing content from welcome.txt
# Reading a file line by line (memory efficient for large files)
with open("introduction.txt", "r") as f:
    for line_num, line in enumerate(f, 1):
        print(f"Line {line_num}: {line.strip()}")
````

## Reading, Modifying & Appending

A common pattern: **read** a file, **process** the data, then **write** or **append** the result.

```python
# Read existing content
with open("data.txt", "r") as f:
    content = f.read()

# Process content
modified = content.upper()

# Append result
with open("data.txt", "a") as f:
    f.write("\n" + modified)
```


````python
# Read the file, modify the content in memory, then append the result
with open("welcome.txt", "r") as readedFile:
    FetchedData = readedFile.read()
    print("Original:\n", FetchedData)

    # Process the data: swap some words/numbers
    FetchedData = FetchedData.replace("Ali", "Ahmad")
    FetchedData = FetchedData.replace("50", "25")
    FetchedData = FetchedData.replace("Data Analysis", "Advance Python")
    print("\nModified:\n", FetchedData)

    # Append the modified copy to the same file
    # NOTE: re-running this cell appends AGAIN — the file grows each time!
    with open("welcome.txt", "a") as File2Write:
        File2Write.write(f"\n {FetchedData}")
````

````python
import json

# Create an empty user database and store it as JSON.
# JSON is a clean, standard format that json.load() can read back reliably.
empty_db = {"Names": [], "Passwd": []}
with open("UsersDatabase.txt", "w") as f:
    json.dump(empty_db, f, indent=2)

print("Database initialized:", empty_db)
````

````python
# NOTE: '1' (str) == 1 (int) is False — always convert input() with int()
choice = int(input("login or signup 1/2 : "))

if choice == 1:  # LOGIN
    mail = input("Enter mail : ")
    passwd = input("Enter passwd : ")

    with open("UsersDatabase.txt", "r") as userFetchedDB:
        db = json.load(userFetchedDB)          # load the JSON back into a dict

    if mail in db["Names"] and passwd in db["Passwd"]:
        print("User logged in")
    else:
        print("User doesn't exist")

elif choice == 2:  # SIGNUP
    mail = input("Enter mail : ")
    passwd = input("Enter passwd : ")

    with open("UsersDatabase.txt", "r") as userFetchedDB:
        db = json.load(userFetchedDB)

    if mail in db["Names"]:
        print("User already exists")
    else:
        db["Names"].append(mail)               # modify the dict in memory
        db["Passwd"].append(passwd)

        with open("UsersDatabase.txt", "w") as newFile2Write:
            json.dump(db, newFile2Write, indent=2)   # save back to disk
        print("Account created and saved!")
````

````python
# json.dump() writes Python objects to file in JSON format
# import the json module for JSON file operations
# Writing JSON with pretty formatting
import json

user_data = {
    "name": "Ahmad",
    "age": 25,
    "skills": ["Python", "NumPy", "Pandas"],
    "address": {
        "city": "Dubai",
        "country": "UAE"
    }
}

# Write with indentation for readability
with open("user.json", "w") as jf:
    json.dump(user_data, jf, indent=4)

# Read and display
with open("user.json", "r") as jf:
    loaded = json.load(jf)
print("Loaded JSON:", json.dumps(loaded, indent=2))
````

## JSON File Handling

JSON (JavaScript Object Notation) is a lightweight data format. Python's `json` module handles it natively.

```python
import json

# Writing
data = {"name": "Ahmad", "age": 25}
with open("data.json", "w") as f:
    json.dump(data, f, indent=4)

# Reading
with open("data.json", "r") as f:
    loaded = json.load(f)
```

- `json.dump(obj, file)` — write to file
- `json.dumps(obj)` — return string
- `json.load(file)` — read from file
- `json.loads(string)` — read from string


````python
import json

# Write to JSON
user_record = {
    "username": "ahmad",
    "email": "ahmad@example.com",
    "active": True,
    "scores": [85, 92, 78]
}

with open("user.json", "w") as jf:
    json.dump(user_record, jf, indent=4)
print("✅ user.json created!")
````

````python
# The result preserves Python data types (dict, list, etc.)
# json.load() reads JSON file and converts back to Python objects
# Read from JSON
with open("user.json", "r") as jf:
    loaded_data = json.load(jf)
    print("Loaded:", loaded_data)
    print("Username:", loaded_data["username"])
    print("Scores:", loaded_data["scores"])
````

## CSV File Handling

Use Python's `csv` module for comma-separated value files.
- `csv.writer(file)` — create a CSV writer
- `csv.reader(file)` — create a CSV reader
- `writer.writerow(list)` — write one row


````python
import csv

# Write a CSV file
with open("students.csv", "w", newline="") as cf:
    writer = csv.writer(cf)
    writer.writerow(["Name", "Age", "Grade"])  # header
    writer.writerow(["Ahmad", 25, "A"])
    writer.writerow(["Ali", 27, "B"])
    writer.writerow(["Sara", 22, "A+"])
print("✅ students.csv created!")
````

````python
# writerow() writes a single row as a list
# csv.writer() creates a writer object for the file
# Read a CSV file
with open("students.csv", "r") as cf:
    reader = csv.reader(cf)
    for row in reader:
        print(row)
````

## Error Handling with Files

Always use `try/except` when working with files to handle:
- `FileNotFoundError` — file doesn't exist
- `PermissionError` — no read/write permission
- `IsADirectoryError` — tried to open a directory as a file
- `UnicodeDecodeError` — wrong text encoding

```python
try:
    with open("data.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("File not found!")
except PermissionError:
    print("Permission denied!")
```


````python
try:
    with open("nonexistent.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("❌ File not found! Please check the filename.")
except PermissionError:
    print("❌ Permission denied! Cannot read the file.")
finally:
    print("Done with file operation.")
````

## OS Module for File Operations

The `os` module provides functions for interacting with the operating system:

| Function | What it does |
|----------|--------------|
| `os.path.exists(p)` | Check if a file/folder exists |
| `os.listdir(d)` | List the contents of a directory |
| `os.rename(a, b)` | Rename / move a file |
| `os.remove(p)` | Delete a file |
| `os.makedirs(p)` | Create directories (including parents) |
| `os.path.join(a, b)` | Join path parts safely |

A quick demo:

````python
import os

# Check if a file exists
print("introduction.txt exists:", os.path.exists("introduction.txt"))

# List everything in the current folder
print("\nFiles in this folder:")
for name in sorted(os.listdir(".")):
    print(" -", name)

# Rename, verify, then delete — clean up after ourselves
if os.path.exists("mode_demo.txt"):
    os.rename("mode_demo.txt", "renamed_demo.txt")
    print("\nRenamed mode_demo.txt -> renamed_demo.txt")
    os.remove("renamed_demo.txt")
    print("And deleted it again.")
````

## Modern Path Handling with pathlib

Python 3.4+ ships with `pathlib` — an object-oriented way to work with paths. It is the **recommended** approach in modern Python:

| Operation | `os` style | `pathlib` style |
|-----------|------------|-----------------|
| Check existence | `os.path.exists(p)` | `Path(p).exists()` |
| Read text file | `open(p).read()` | `Path(p).read_text()` |
| Write text file | `open(p, "w").write(t)` | `Path(p).write_text(t)` |
| List a directory | `os.listdir(d)` | `Path(d).iterdir()` |
| Join paths | `os.path.join(a, b)` | `a / b` |

````python
from pathlib import Path

# One object holds the path AND convenient methods
intro = Path("introduction.txt")
print("Exists?", intro.exists())
print(intro.read_text())          # read the whole file in one line

# Write a brand-new file
log = Path("pathlib_demo.txt")
log.write_text("Created with pathlib!\n")
print("Created:", log.read_text())

# List every .txt file in this folder with glob()
print("\nAll .txt files here:")
for p in sorted(Path(".").glob("*.txt")):
    print(" -", p.name)
````

## 🎯 Key Takeaways

- `open(path, mode)` + the `with` statement is the safe way to do file I/O — the file is always closed, even on errors.
- `"w"` **overwrites**, `"a"` **appends**, `"r"` **reads** — pick the right mode.
- `f.read()` reads everything; iterating the file object reads line by line (memory-friendly for big files).
- The `json` module turns Python dicts into files and back — perfect for configuration and small databases.
- The `csv` module handles CSV files with `writer.writerow()` / `csv.reader()`.
- Wrap file access in `try/except` to survive missing files and permission errors.
- `pathlib.Path` is the modern, readable way to work with paths.
