# File Handling in Python

Python can read from and write to files on disk. The `open()` function is used to open a file and returns a file object. The `with` statement ensures the file is properly closed after use.

**File modes:**
- `"r"` — read (default)
- `"w"` — write (overwrites existing file)
- `"a"` — append (adds to end)
- `"r+"` — read + write
- `"b"` — binary mode (e.g., `"rb"`, `"wb"`)


### Why File Handling Matters

Being able to **read from** and **write to** files is essential for real-world programming. You'll need it for:

- Saving program output and logs
- Loading configuration files
- Processing datasets (CSV, JSON, Excel)
- Reading/writing user databases
- Interacting with other applications

Python's `open()` function and the `with` statement make file I/O clean and safe.


## Writing to Files

Use `open("filename", "w")` with `file.write()` to write content.
**Warning:** `"w"` mode overwrites the file if it already exists!


# file.write() writes a string to the file
# 'w' mode creates the file if it doesn't exist, overwrites if it does
### File Modes Explained

The second argument to `open()` specifies the **mode**:

| Mode | Description |
|------|-------------|
| `"r"`  | Read (default) — file must exist |
| `"w"`  | Write — **creates/overwrites** file |
| `"a"`  | Append — adds to the end of existing file |
| `"x"`  | Exclusive creation — fails if file exists |
| `"r+"` | Read + Write (no truncation) |
| `"w+"` | Read + Write (truncates file) |
| `"a+"` | Read + Append |
| `"b"`  | Binary mode (add to any mode: `"rb"`, `"wb"`) |

**Always** use the `with` statement — it automatically closes the file, even if an error occurs.


````python
msg="Ahmad here !"
with open("introduction.txt","w") as file:
    file.write(msg)
````

````python
# .read() reads the entire file content into a string
# 'r' mode opens the file for reading (default)
# Writing multiple lines at once
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
with open("multi_line.txt", "w") as f:
    f.writelines(lines)

# Verify by reading
with open("multi_line.txt", "r") as f:
    print("File contents:")
    print(f.read())
````

````python
# Demonstrate different file modes
# 'w' — write mode (creates new file or overwrites existing)
with open("mode_demo.txt", "w") as f:
    f.write("This was written in 'w' mode.\n")

# 'a' — append mode (adds to existing content)
with open("mode_demo.txt", "a") as f:
    f.write("This was appended in 'a' mode.\n")

# 'r' — read mode
with open("mode_demo.txt", "r") as f:
    content = f.read()
    print("Content:", content)
````

## Reading, Modifying & Appending

- `file.read()` — reads entire file as a single string
- `file.readline()` — reads one line at a time

## JSON File Handling

JSON is a lightweight data format. Python's `json` module handles it natively.
- `json.dump(obj, file)` — write object to file
- `json.load(file)` — read object from file


````python
# json.dump() writes Python objects to file in JSON format
# import the json module for JSON file operations
# Writing JSON with pretty formatting
import json

user_data = {
    "name": "Ahmad",
    "age": 25,
    "skills": ["Python", "NumPy", "Pandas"],
    "address": {
        "city": "Dubai",
        "country": "UAE"
    }
}

# Write with indentation for readability
with open("user.json", "w") as jf:
    json.dump(user_data, jf, indent=4)

# Read and display
with open("user.json", "r") as jf:
    loaded = json.load(jf)
print("Loaded JSON:", json.dumps(loaded, indent=2))
````

## CSV File Handling

Use Python's `csv` module for comma-separated value files.
- `csv.writer(file)` — create a CSV writer
- `csv.reader(file)` — create a CSV reader
- `writer.writerow(list)` — write one row


````python
# import csv module for comma-separated value files
# Appending new rows to an existing CSV
import csv

# Append a new student to the CSV
new_student = {"Name": "Moiz", "Age": 22, "City": "Dubai"}

with open("students.csv", "a", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["Name", "Age", "City"])
    writer.writerow(new_student)

# Verify the appended data
print("All students after append:")
with open("students.csv", "r") as f:
    for row in csv.DictReader(f):
        print(f"  {row['Name']:10} {row['Age']:3} {row['City']}")
````

````python
# Using DictWriter and DictReader (more readable than plain csv.writer)
import csv

# Write with DictWriter
fieldnames = ["Name", "Age", "City"]
with open("students.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"Name": "Ahmad", "Age": 25, "City": "Dubai"})
    writer.writerow({"Name": "Sarah", "Age": 23, "City": "Abu Dhabi"})
    writer.writerow({"Name": "Ali", "Age": 22, "City": "Sharjah"})

# Read with DictReader
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['Name']} is {row['Age']} years old from {row['City']}")
````

## Error Handling with Files

Always use `try/except` when working with files to handle missing files, permission errors, etc.


````python
# try/except catches FileNotFoundError gracefully
# Robust file handling with try/except
import os

def safe_read_file(filename):
    """Safely read a file with proper error handling."""
    try:
        with open(filename, "r") as f:
            return f.read()
    except FileNotFoundError:
        return f"Error: '{filename}' not found."
    except PermissionError:
        return f"Error: Permission denied for '{filename}'."
    except IOError as e:
        return f"Error reading '{filename}': {e}"

# Test with an existing file
print("Existing file:")
print(safe_read_file("introduction.txt"))

# Test with a non-existent file
print("\nMissing file:")
print(safe_read_file("does_not_exist.txt"))
````

## Error Handling with Files

Always use `try/except` when working with files to handle:
- `FileNotFoundError` — file doesn't exist
- `PermissionError` — no read/write permission
- `IsADirectoryError` — tried to open a directory as a file
- `UnicodeDecodeError` — wrong text encoding

```python
try:
    with open("data.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("File not found!")
except PermissionError:
    print("Permission denied!")
```

## OS Module for File Operations

The `os` module provides functions for interacting with the operating system.


````python
# os.remove() deletes a file from the filesystem
# import os module for operating system interactions
# shutil — high-level file operations (copy, move, delete)
import shutil
import os

# Copy a file
if os.path.exists("introduction.txt"):
    shutil.copy("introduction.txt", "introduction_backup.txt")
    print("Copied introduction.txt → introduction_backup.txt")

# Read the copy to verify
with open("introduction_backup.txt", "r") as f:
    print("Backup content:", f.read())
````

````python
# Working with binary files
# Write binary data
binary_data = bytes([0, 1, 2, 3, 255, 254, 253])
with open("binary_demo.bin", "wb") as f:
    f.write(binary_data)

# Read binary data
with open("binary_demo.bin", "rb") as f:
    read_data = f.read()
print("Written:", list(binary_data))
print("Read:", list(read_data))
print("Match:", binary_data == read_data)
````

````python
# Modern pathlib approach
from pathlib import Path

# Create a Path object
p = Path("introduction.txt")

print(f"Path: {p}")
print(f"Exists: {p.exists()}")
print(f"Is file: {p.is_file()}")
print(f"Name: {p.name}")
print(f"Stem (name without extension): {p.stem}")
print(f"Suffix (extension): {p.suffix}")
print(f"Parent directory: {p.parent}")
print(f"Absolute path: {p.absolute()}")

# Read content using pathlib
if p.exists():
    content = p.read_text()
    print(f"\nContent:\n{content}")
````
