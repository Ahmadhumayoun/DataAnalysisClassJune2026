```python
computerGuessedNumber = 57
userGuessedNumber = int(input("Enter your guessed number : "))
chances = 3
while chances >= 0:
    if userGuessedNumber == computerGuessedNumber:
        print("You guessed the number correctly")
        break
    elif userGuessedNumber > computerGuessedNumber:
        print("Your guessed number is greater than the computer guessed number")
    else:
        print("Your guessed number is less than the computer guessed number")
    chances = chances - 1
    print(f"You have {chances} chances left")
    userGuessedNumber = int(input("Enter your guessed number : "))
```
