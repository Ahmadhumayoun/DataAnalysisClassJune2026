### Why NumPy is Faster Than Python Lists

| Feature | Python List | NumPy Array |
|---------|-------------|-------------|
| Memory | Stores objects (pointers) | Stores values in contiguous memory |
| Operations | Must loop element-by-element | Vectorized (C-level loops) |
| Data types | Mixed types allowed | Homogeneous (all same type) |
| Speed | Slow for numerical data | 10-100x faster |

NumPy achieves this by using **fixed-type data** stored in a **contiguous memory block**, enabling highly optimized C and Fortran code to operate on the data.

## Boolean Indexing (Filtering)

One of NumPy's most powerful features: apply a condition directly to an array to get a boolean mask, then use that mask to select elements.

```python
arr = np.array([1, 2, 3, 4, 5])
```

## Array Operations & Broadcasting

NumPy performs element-wise operations without loops (vectorization). Broadcasting automatically expands smaller arrays to match larger ones.


````python
# Each element is added element-wise
# Vectorized operations — no explicit loop needed!
# Concatenating arrays
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])

v_stack = np.vstack((a, b))  # vertical stack (add rows)
print("Vstack:\n", v_stack)

c = np.array([[5], [6]])
h_stack = np.hstack((a, c))  # horizontal stack (add columns)
print("\nHstack:\n", h_stack)
````

````python
# Dot product (matrix multiplication)
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

dot_product = np.dot(A, B)   # matrix multiplication
print("A dot B:\n", dot_product)

# @ operator — same as np.dot() (Python 3.5+)
print("\nA @ B:\n", A @ B)

# Element-wise multiplication (not matrix multiplication!)
print("\nA * B (element-wise):\n", A * B)
````

````python
# Broadcasting example: add a scalar to an array
arr = np.array([1, 2, 3, 4])
print("arr + 10:", arr + 10)   # [11, 12, 13, 14]

# Broadcasting: add a 1D array to a 2D array
matrix = np.array([[1, 2, 3], [4, 5, 6]])
row = np.array([10, 20, 30])
print("\nMatrix + row:\n", matrix + row)  # row broadcasts to match matrix

# Broadcasting: different shapes
A = np.array([[1], [2], [3]])  # shape (3, 1)
B = np.array([10, 20, 30])    # shape (3,)
print("\nA (3,1) + B (3,):\n", A + B)  # result (3, 3)
````

## Broadcasting Rules

**Broadcasting** lets NumPy perform operations on arrays of different shapes. NumPy automatically "stretches" the smaller array to match the larger one.

**Rules:**
1. If arrays have different dimensions, prepend 1s to the smaller shape
2. Arrays are compatible if dimensions are equal or one is 1
3. The result takes the larger shape

**Examples:**
- `(3, 4) + (4,)` → `(3, 4) + (1, 4)` → `(3, 4)` ✓
- `(3, 4) + (3,)` → `(3, 4) + (1, 3)` → ✗ (different trailing dims)
- `(3, 1) + (1, 4)` → `(3, 4)` ✓ (both dimensions broadcast)


````python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print(a + b)    # [11 22 33 44] — element-wise addition
print(a * b)    # [ 10  40  90 160] — element-wise multiplication
print(a ** 2)   # [ 1  4  9 16] — square each element
print(a * 10)   # [10 20 30 40] — broadcasting: scalar × array
````

````python
# Student marks for grading demonstration
studentMarks = np.array([25 , 50, 80 , 90 ,10 , 33])
# 80 > A
# 80 < B
# 50 < C
# 33 <= F 
````

````python
# Manual grading with Python loop
gradeList = []
for i in studentMarks:
    if i >=80:
        gradeList.append("A")
    elif i<80 and i >= 50:
        gradeList.append("B")
    elif i>= 33 and i < 50:
        gradeList.append("C")
    elif i<=33:
        gradeList.append("F")
gradeArray = np.array(gradeList)
````

````python
# Display original marks and computed grades
print(studentMarks)
print(gradeArray)
````

````python
# Much faster than a Python for loop!
# Vectorized grading using np.select
studentMarks = np.array([25 , 50, 80 , 90 ,10 , 33])

condition = [(studentMarks >=80),(studentMarks >=50),(studentMarks>=33)]
grade = ["A","B","C"]
gradeArray = np.select(condition,grade,default="F")
print(gradeArray)
````

## Statistical Functions

NumPy provides many statistical functions that operate on entire arrays:
- `np.mean(arr)`, `np.median(arr)`, `np.std(arr)`
- `np.min(arr)`, `np.max(arr)`, `np.sum(arr)`
- Use `axis=0` for columns, `axis=1` for rows


````python
# Statistical summary of a dataset
# Statistical functions with axis parameter
data_2d = np.array([[10, 20, 30], [40, 50, 60]])

print("Matrix:\n", data_2d)
print("Total sum:", np.sum(data_2d))
print("Column sums (axis=0):", np.sum(data_2d, axis=0))
print("Row sums (axis=1):", np.sum(data_2d, axis=1))
print("Mean of all:", np.mean(data_2d))
print("Std dev:", np.std(data_2d))
print("Percentile (50th / median):", np.percentile(data_2d, 50))
````

### Axis Parameter

The `axis` parameter controls **which dimension** the operation applies to:

- `axis=0` — operate **down rows** (vertically, column-wise)
- `axis=1` — operate **across columns** (horizontally, row-wise)
- No axis — operate on **all elements** (flattened)

```python
matrix = np.array([[1, 2], [3, 4]])
np.sum(matrix)        # 10  (total)
np.sum(matrix, 0)     # [4, 6]  (column sums)
np.sum(matrix, 1)     # [3, 7]  (row sums)
```


````python
data = np.array([12, 45, 23, 67, 34, 89, 15, 42])
print(f"Mean:   {np.mean(data):.2f}")
print(f"Median: {np.median(data):.2f}")
print(f"Std:    {np.std(data):.2f}")
print(f"Min:    {np.min(data)}")
print(f"Max:    {np.max(data)}")
print(f"Sum:    {np.sum(data)}")

# Axis example with 2D array
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Row sums:", np.sum(matrix, axis=1))    # [ 6 15]
print("Col sums:", np.sum(matrix, axis=0))    # [5 7 9]
````

## Random Module

`np.random` generates random numbers for simulations and testing:
- `np.random.randint(low, high, size)` — random integers
- `np.random.rand(d0, d1, ...)` — uniform [0, 1)
- `np.random.randn(d0, d1, ...)` — standard normal distribution


````python
# randint generates random integers in [low, high)
# Seed ensures reproducible random numbers
# Different random distributions
np.random.seed(123)

# Uniform distribution [0, 1)
uniform = np.random.rand(1000)
print("Uniform mean:", uniform.mean(), "(expected ~0.5)")

# Standard normal distribution (mean=0, std=1)
normal = np.random.randn(1000)
print("Normal mean:", normal.mean(), "std:", normal.std())

# Random integers
rand_ints = np.random.randint(1, 7, 10)  # like rolling a die 10 times
print("Die rolls:", rand_ints)

# Normal distribution with custom parameters
custom_normal = np.random.normal(loc=100, scale=15, size=5)
print("Normal(100, 15):", custom_normal)
````
