- Method overriding and `super()`

## Functions — The Building Blocks

A **function** is a reusable block of code with a name, optional parameters, and an optional `return` value. Once defined, you call it as many times as you like.

````python
def convey_message(your_name):
    print(f"Hello, {your_name}!")
    print("Welcome to Python Class")
    print("This is a simple function to demonstrate the concept of functions in Python.")
````

````python
# Call the function with different names
convey_message("Farwa")
convey_message("Ahmad")
````

Functions can also decide what to return based on their inputs — like this mini calculator:

````python
def calculator(operation, num1, num2):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "m":
        return num1 * num2
    elif operation == "divide":
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero is not allowed."
    else:
        return "Error: Invalid operation. Please choose from 'add', 'subtract', 'multiply', or 'divide'."
````

````python
# "m" stands for multiply
print(calculator("m", 10, 5))
print(calculator("divide", 10, 5))
print(calculator("divide", 10, 0))  # handled gracefully
````

A function sends a value back to the caller with `return`:

````python
def testReturn():
    return ("Ahmad")
````

````python
name = testReturn()
print(name+" Humayoun")
````

## Classes and Objects — The OOP Core

A **class** is a blueprint; an **object** is a concrete instance built from that blueprint. Here `vehicle` bundles the parts of a car into one object:

````python
class vehicle:

    def wheel(self):
        return("Wheel are used in vehicles to move from one place to another.")
    def body(self):
        return("Body is the outer part of the vehicle which protects the passengers and other components.")

    def engine(self,carName):
        return(f"Engine is the heart of the {carName} which provides power to move the vehicle.")
````

````python
v = vehicle()
print(v.engine("toyota"))
print(v.wheel())
print(v.body())
````

### Real-World Example: Cleaning Spreadsheet Prices

This class solves the exact problem we hit in the Pandas notebooks — prices stored as messy strings like `"100,00"`:

````python


dirtPrice = ["100,00","200,00"]
class SheetAnalysis:
    def cleanPrice(self,ListToclean,elementToRemove):
        cleanedList = []
        for price in ListToclean:
            cleanedPrice = price.replace(elementToRemove, "")
            cleanedList.append(cleanedPrice)
        return cleanedList
````

````python
sa = SheetAnalysis()
cleanedPrices = sa.cleanPrice(dirtPrice, ",")
print(cleanedPrices)
````

## Methods that Change State

Methods can read *and* update the object's attributes. `deposit()` permanently changes the balance — state lives inside the object:

````python
class bankAccount:
    def __init__(self, balance):
        self.balance = balance

    def checkBalance(self):
        return self.balance
    def deposit(self, amount):
        self.balance += amount
        return self.balance
        
````

````python
b = bankAccount(5000)
print(b.checkBalance())
b.deposit(2000)
print(b.checkBalance())
````

A class with no constructor still works — Python gives every object a default one:

````python
class ABC :
    def displayData(self):
        print("This is a method inside class ABC.")

        
````

````python
a = ABC()
a.displayData()
````

## Inheritance

**Inheritance** lets a child class reuse everything from a parent class — `ChildClass` gets all of its methods for free. Grandparent → parent → child chains work too:

````python
class grandparentClass:

    def functionFromParent(self):
        print("This is a method from the parent class.")

class parentClass(grandparentClass):
    def functionFromChild(self):
        print("This is a method from the child class.")
class ChildClass(parentClass):
    pass


c = ChildClass()
c.functionFromChild()
c.functionFromParent()
````

### Multiple Inheritance

A class can inherit from **several** parents at once:

````python
class parent1:
    def functionFromParent1(self):
        print("This is a method from the parent class 1.")

class parent2:
    def functionFromParent2(self):
        print("This is a method from the parent class 2.")

class hybridChildClass(parent1, parent2):
    pass

hyb = hybridChildClass()
hyb.functionFromParent1()
hyb.functionFromParent2()
````

### Method Overriding

A child can **override** a parent method by redefining it with the same name — the child's version wins:

````python
class parentClass:

    def displayData(self):
        print("This is a method from the parent class.")

class childClass(parentClass):
    def displayData(self):
        print("This is a method from the child class.")

c = childClass()
c.displayData()
````

### `super()` — Calling the Parent Constructor

When the child adds new attributes, call `super().__init__(...)` to initialize the parent's part first. This cell originally crashed with `AttributeError: 'childClass' object has no attribute 'displayAge'` — the fix is the `displayAge` method below:

````python
class parentClass:
    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

    def displayName(self):
        print(f"Name: {self.name}")

class childClass(parentClass):
    def __init__(self, name, gender, age):
        super().__init__(name, gender)   # parent handles name & gender
        self.age = age                   # child adds its own attribute

    def displayName(self):
        print(f"Name: {self.name}")

    def displayAge(self):
        print(f"Age: {self.age}")

child = childClass("Ahmad", "Male", 25)
child.displayName()
child.displayAge()
````

## Access Modifiers in Python

Python has **conventions**, not strict enforcement:

| Prefix | Name | Meaning |
|--------|------|---------|
| `method()` | public | accessible everywhere |
| `_method()` | protected | "internal use" — still callable |
| `__method()` | private | name-mangled, discouraged from outside |

````python
class parentClass:
    

    def publicFunction(self):
        return"this is a public function"

    def __privateFunction(self):
        return "this is private function"
        
    def _protectedFunction(self):
        return "This is a protected Function"
class childClass(parentClass):

    def displayProtectedFunction(self):
        return str(self._protectedFunction())

c = childClass()
c.displayProtectedFunction()
````

## Capstone Project: A Bank in OOP

We now combine **everything**: two classes (`Account` = one customer, `Bank` = many accounts stored in a pandas DataFrame backed by `bank.csv`). Run the cells, then use the menu to create/login/withdraw — this is a *real* interactive program:

````python
import pandas as pd

class Account:
    def __init__(self, accountNumber, fname, lname, password, address, studentClass, city, balance):
        self.accountNumber = accountNumber
        self.fname = fname
        self.lname = lname
        self.password = password
        self.address = address
        self.studentClass = studentClass
        self.city = city
        self.balance = balance

    def display(self):
        print("First Name :", self.fname)
        print("Last Name :", self.lname)
        print("Address :", self.address)
        print("Class :", self.studentClass)
        print("City :", self.city)
        print("Balance :", self.balance)

    def checkBalance(self):
        return self.balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return self.balance
        return None
````

````python
class Bank:
    def __init__(self, filename="../03_Pandas_Data_Analysis/bank.csv"):
        self.filename = filename
        self.accounts = {}
        self.load()

    def load(self):
        df = pd.read_csv(self.filename)
        for i in range(len(df)):
            account = str(df["AccountNumber"][i])
            self.accounts[account] = Account(
                account,
                df["FName"][i],
                df["LName"][i],
                str(df["Password"][i]),
                df["Address"][i],
                df["Class"][i],
                df["City"][i],
                float(df["Balance"][i])
            )

    def save(self):
        rows = []
        for account in self.accounts:
            acc = self.accounts[account]
            rows.append([acc.accountNumber, acc.fname, acc.lname, acc.password,
                         acc.address, acc.studentClass, acc.city, acc.balance])
        df = pd.DataFrame(rows, columns=["AccountNumber", "FName", "LName",
                                         "Password", "Address", "Class", "City", "Balance"])
        df.to_csv(self.filename, index=False)
        print("Data Saved")

    def createAccount(self):
        account = input("Account Number: ")
        if account in self.accounts:
            print("Account Already Exists")
        else:
            fname = input("First Name: ")
            lname = input("Last Name: ")
            password = input("Password: ")
            address = input("Address: ")
            studentClass = input("Class: ")
            city = input("City: ")
            balance = float(input("Opening Balance: "))
            self.accounts[account] = Account(account, fname, lname, password,
                                             address, studentClass, city, balance)
            print("Account Created")

    def login(self):
        account = input("Account Number: ")
        password = input("Password: ")
        if account in self.accounts and password == self.accounts[account].password:
            self.accountMenu(account)
        else:
            print("Invalid Account or Password")

    def accountMenu(self, account):
        while True:
            print("\n1. View Details")
            print("2. Check Balance")
            print("3. Deposit")
            print("4. Withdraw")
            print("5. Logout")
            option = input("Choice: ")
            acc = self.accounts[account]
            if option == "1":
                acc.display()
            elif option == "2":
                print("Balance :", acc.checkBalance())
            elif option == "3":
                amount = float(input("Amount: "))
                print("New Balance :", acc.deposit(amount))
            elif option == "4":
                amount = float(input("Amount: "))
                result = acc.withdraw(amount)
                if result is None:
                    print("Insufficient Balance")
                else:
                    print("Remaining Balance :", result)
            elif option == "5":
                break
            else:
                print("Invalid Choice")
````

> **Interactive cell** — run it and pick menu options:
>
> ```
> 1. Create Account      → add a new customer (stored in bank.csv on exit)
> 2. Login              → pick an existing account and operate on it
> 3. Exit               → save changes back to bank.csv
> ```

````python
def main_menu(bank):
    while True:
        print("\n1. Create Account")
        print("2. Login")
        print("3. Exit")
        choice = input("Enter Choice: ")
        if choice == "1":
            bank.createAccount()
        elif choice == "2":
            bank.login()
        elif choice == "3":
            bank.save()
            break
        else:
            print("Invalid Choice")

bank = Bank()
main_menu(bank)
````

## 🎯 Key Takeaways

- A **class** is a blueprint; an **object** is an instance of it.
- `__init__` is the constructor — it sets up the object's attributes.
- `self` always refers to *the current object*.
- Methods can read and **mutate** state (`deposit` changes `self.balance`).
- **Inheritance** reuses parent code; overriding replaces it; `super()` initializes the parent's part.
- `_protected` and `__private` are conventions — Python trusts you.
- OOP shines in bigger systems: one `Bank` object managing many `Account` objects.
