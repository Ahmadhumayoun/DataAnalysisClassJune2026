## Reading Excel Files

`pd.read_excel()` loads Excel (`.xlsx`, `.xls`) files. Specify `sheet_name` to read a specific sheet.

```python
pd.read_excel("file.xlsx", sheet_name="Sheet1", header=0, usecols="A:C")
```

- `sheet_name`: name or index (0-based) of the sheet to read
- `header`: which row holds the column names (default 0)
- `usecols`: which columns to read
- `dtype`: force column types

Each sheet becomes a separate DataFrame — read them all by calling `pd.read_excel()` once per sheet.

````python
# Read the "Items" sheet into a DataFrame
itemsDf = pd.read_excel("VendorProducts.xlsx", sheet_name="Items")

# Read the "Collections" sheet into a separate DataFrame
collectionsDf = pd.read_excel("VendorProducts.xlsx", sheet_name="Collections")
````

## Project: Banking System with CSV

This project demonstrates a full banking system using pandas for data storage. It loads account data from `bank.csv`, allows account creation and login, and saves changes back to the CSV file on exit.

**Features:** Create account, Login, View details, Check balance, Deposit, Withdraw, Save to CSV

> **Note:** this cell uses `input()` — run it and interact with the menu. Press **3** to exit, which saves all changes back to `bank.csv`.

````python
# BANK SYSTEM — DO NOT MODIFY
# Read bank CSV into dictionary for in-memory manipulation
df = pd.read_csv("bank.csv")
bank = {}
for i in range(len(df)):
    account = str(df["AccountNumber"][i])
    bank[account] = {
        "FName": df["FName"][i],
        "LName": df["LName"][i],
        "Password": str(df["Password"][i]),
        "Address": df["Address"][i],
        "Class": df["Class"][i],
        "City": df["City"][i],
        "Balance": float(df["Balance"][i])
    }
while True:
    print("\n1. Create Account")
    print("2. Login")
    print("3. Exit")
    choice = input("Enter Choice: ")
    if choice == "1":
        account = input("Account Number: ")
        if account in bank:
            print("Account Already Exists")
        else:
            bank[account] = {}
            bank[account]["FName"] = input("First Name: ")
            bank[account]["LName"] = input("Last Name: ")
            bank[account]["Password"] = input("Password: ")
            bank[account]["Address"] = input("Address: ")
            bank[account]["Class"] = input("Class: ")
            bank[account]["City"] = input("City: ")
            bank[account]["Balance"] = float(input("Opening Balance: "))
            print("Account Created")

    elif choice == "2":
        account = input("Account Number: ")
        password = input("Password: ")
        if account in bank and password == bank[account]["Password"]:
            while True:
                print("\n1. View Details")
                print("2. Check Balance")
                print("3. Deposit")
                print("4. Withdraw")
                print("5. Logout")
                option = input("Choice: ")
                if option == "1":
                    print("First Name :", bank[account]["FName"])
                    print("Last Name :", bank[account]["LName"])
                    print("Address :", bank[account]["Address"])
                    print("Class :", bank[account]["Class"])
                    print("City :", bank[account]["City"])
                    print("Balance :", bank[account]["Balance"])
                elif option == "2":
                    print("Balance :", bank[account]["Balance"])
                elif option == "3":
                    amount = float(input("Amount: "))
                    bank[account]["Balance"] = bank[account]["Balance"] + amount
                    print("New Balance :", bank[account]["Balance"])
                elif option == "4":
                    amount = float(input("Amount: "))
                    if amount <= bank[account]["Balance"]:
                        bank[account]["Balance"] = bank[account]["Balance"] - amount
                        print("Remaining Balance :", bank[account]["Balance"])
                    else:
                        print("Insufficient Balance")
                elif option == "5":
                    break
                else:
                    print("Invalid Choice")
        else:
            print("Invalid Account or Password")
    elif choice == "3":
        accounts = []
        fname = []
        lname = []
        password = []
        address = []
        student_class = []
        city = []
        balance = []
        for account in bank:
            accounts.append(account)
            fname.append(bank[account]["FName"])
            lname.append(bank[account]["LName"])
            password.append(bank[account]["Password"])
            address.append(bank[account]["Address"])
            student_class.append(bank[account]["Class"])
            city.append(bank[account]["City"])
            balance.append(bank[account]["Balance"])
        df = pd.DataFrame({
            "AccountNumber": accounts,
            "FName": fname,
            "LName": lname,
            "Password": password,
            "Address": address,
            "Class": student_class,
            "City": city,
            "Balance": balance
        })
        df.to_csv("bank.csv", index=False)
        print("Data Saved")
        break
    else:
        print("Invalid Choice")
````

## Row Filtering & Selection

Filter rows with boolean conditions:

```python
# Single condition
df[df["age"] > 20]

# Multiple conditions (use & for and, | for or)
df[(df["age"] > 20) & (df["age"] < 30)]

# Using .query() for cleaner syntax
df.query("age > 20 and age < 30")

# Using .loc with conditions
df.loc[df["age"] > 20, ["name", "age"]]
```

**Note:** use parentheses around each condition — `&` and `|` bind tighter than comparison operators.

````python
# Boolean filtering: the condition inside [] selects matching rows
print("Students older than 20:")
print(data[data["age"] > 20])

print("\nStudents aged between 20 and 30:")
print(data[(data["age"] >= 20) & (data["age"] <= 30)])

# .loc lets us filter rows AND pick columns at the same time
print("\nOnly the names of students older than 20:")
print(data.loc[data["age"] > 20, "Name"])
````

## Handling Missing Data

Real-world data often has missing values, shown as `NaN`. Here are the key tools:

| Method | Description |
|--------|-------------|
| `.isna()` / `.isnull()` | Detect missing values |
| `.notna()` / `.notnull()` | Detect non-missing values |
| `.dropna()` | Remove rows/columns with NaN |
| `.fillna(value)` | Replace NaN with a given value |
| `.fillna(method="ffill")` | Forward-fill the last valid value |
| `.interpolate()` | Fill with interpolated values

````python
# .sum() counts True values per column
# .isna() returns True for NaN values
# Demonstrate missing data handling
df_with_nan = pd.DataFrame({
    "A": [1, 2, None, 4],
    "B": [None, 2, 3, 4],
    "C": [1, None, None, 4]
})

print("Original:\n", df_with_nan)
print("\nIs NaN?:\n", df_with_nan.isna())
print("\nFill NaN with 0:\n", df_with_nan.fillna(0))
print("\nDrop rows with any NaN:\n", df_with_nan.dropna())
print("\nForward fill:\n", df_with_nan.fillna(method="ffill"))
````

## Data Cleaning: Preparing the Price Column

Real-world data is rarely clean. In `VendorProducts.xlsx` the price column is a **string** like `"22,500د.ع"` (Iraqi Dinar with a thousands separator). Before we can sort, group, or compute statistics, we must convert it to numbers.

We use the **vectorized** `.str` accessor — pandas applies the operation to every value at once, no Python loop needed:

````python
# Step 1: strip the thousands separator "," and the currency suffix "د.ع"
# Step 2: convert the cleaned strings to numbers (errors="coerce" -> NaN for unparseable)
itemsDf["Price"] = pd.to_numeric(
    itemsDf["Price"].astype(str).str.replace(",", "").str.replace("د.ع", ""),
    errors="coerce"
)

# Verify: describe() now reports real statistics
itemsDf["Price"].describe()
````

## Sorting & Grouping

**Sorting:**
```python
df.sort_values(by="column", ascending=False)
df.sort_index()
```

**Grouping** — split data into groups, apply a function, combine the results:
```python
df.groupby("category")["value"].mean()
df.groupby(["cat1", "cat2"]).agg({"val1": "mean", "val2": "sum"})
```

Grouping is one of the most powerful pandas operations — it is like pivot tables in Excel.

````python
# Sort students by age (ascending by default)
data.sort_values(by="age")

# Sort the products from most to least expensive
itemsDf.sort_values(by="Price", ascending=False)[["Name", "Price"]].head(10)
````

````python
# Average price by brand — groupby splits, aggregates, and combines
itemsDf.groupby("Brand")["Price"].agg(["count", "mean", "min", "max"]).head(10)
````

````python
# Price column likely has currency symbols that need cleaning
# Sorting requires clean numeric data
# Pivot table — like Excel pivot tables
if "Category" in itemsDf.columns and "Brand" in itemsDf.columns:
    pivot = pd.pivot_table(
        itemsDf,
        values="Price",
        index="Category",
        columns="Brand",
        aggfunc="mean",
        fill_value=0
    )
    print("Pivot Table (Avg Price by Category & Brand):\n", pivot)
````

## Merging & Concatenating

Combine multiple DataFrames:

```python
# Concatenate (stack) DataFrames
pd.concat([df1, df2])          # vertical stack (more rows)
pd.concat([df1, df2], axis=1)  # horizontal stack (more columns)

# Merge (like SQL JOIN)
pd.merge(df1, df2, on="key_column")
pd.merge(df1, df2, left_on="id", right_on="student_id", how="left")
```

**Merge types:** `inner`, `outer`, `left`, `right` (same as SQL).

````python
# Create a simple grades DataFrame
grades_df = pd.DataFrame({
    "Name": ["Ahmad", "Ali", "Moiz"],
    "Grade": ["A", "B", "A"],
    "Subject": ["Math", "Math", "Science"]
})

print("Student data:\n", data)
print("\nGrades:\n", grades_df)

# Merge on the shared "Name" column — like a SQL LEFT JOIN
merged = pd.merge(data, grades_df, on="Name", how="left")
print("\nMerged data:\n", merged)
````

## 🎯 Key Takeaways

- Pandas provides two core structures: **Series** (1D labeled) and **DataFrame** (2D labeled table).
