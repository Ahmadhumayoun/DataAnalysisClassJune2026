## Row Filtering & Selection

Filter rows using boolean conditions:
- `df[df["column"] > value]` — conditional filtering
- `df.loc[row_label, col_label]` — label-based selection
- `df.iloc[row_index, col_index]` — integer position-based selection


````python
# Boolean filtering: condition inside [] selects matching rows
# Filtering with multiple conditions
# Using the student DataFrame
print("Students aged between 20 and 30:")
print(data[(data["age"] >= 20) & (data["age"] <= 30)])

print("\nStudents named Ahmad (case-insensitive):")
# Use .str accessor for string operations
if data["Name"].dtype == object:
    print(data[data["Name"].str.contains("Ahmad", case=False)])
````

## Row Filtering & Selection

Filter rows with boolean conditions:

```python
# Single condition
df[df["age"] > 20]

# Multiple conditions (use & for and, | for or)
df[(df["age"] > 20) & (df["age"] < 30)]

# Using .query() for cleaner syntax
df.query("age > 20 and age < 30")

# Using .loc with conditions
df.loc[df["age"] > 20, ["name", "age"]]
```

**Note:** Use parentheses around each condition — `&` and `|` have higher precedence than comparison operators.


````python
# Filtering example with the student DataFrame
print("Students older than 20:")
print(data[data["age"] > 20])
print("\nNames of students older than 20:")
print(data[data["age"] > 20]["Name"])
````

## Handling Missing Data

- `.dropna()` — remove rows with NaN
- `.fillna(value)` — fill NaN with a value
- `.isna()` / `.notna()` — detect missing values


````python
# .sum() counts True values per column
# .isna() returns True for NaN values
# Demonstrate missing data handling
df_with_nan = pd.DataFrame({
    "A": [1, 2, None, 4],
    "B": [None, 2, 3, 4],
    "C": [1, None, None, 4]
})

print("Original:\n", df_with_nan)
print("\nIs NaN?:\n", df_with_nan.isna())
print("\nFill NaN with 0:\n", df_with_nan.fillna(0))
print("\nDrop rows with any NaN:\n", df_with_nan.dropna())
print("\nForward fill:\n", df_with_nan.fillna(method="ffill"))
````

## Handling Missing Data

Real-world data often has missing values (shown as `NaN`).

| Method | Description |
|--------|-------------|
| `.isna()` / `.isnull()` | Detect missing values |
| `.notna()` / `.notnull()` | Detect non-missing values |
| `.dropna()` | Remove rows/columns with NaN |

| `.fillna(method="ffill")` | Forward-fill last valid value |
| `.interpolate()` | Fill with interpolated values |

**Tip:** Use `inplace=True` to modify the DataFrame directly.

## Sorting & Grouping

- `.sort_values(by="column")` — sort DataFrame
- `.groupby("column").agg({"col": "func"})` — group and aggregate


````python
# Price column likely has currency symbols that need cleaning
# Sorting requires clean numeric data
# Pivot table — like Excel pivot tables
if "Category" in itemsDf.columns and "Brand" in itemsDf.columns:
    pivot = pd.pivot_table(
        itemsDf,
        values="Price",
        index="Category",
        columns="Brand",
        aggfunc="mean",
        fill_value=0
    )
    print("Pivot Table (Avg Price by Category & Brand):\n", pivot)
````

````python
# GroupBy examples with the items DataFrame
if "Category" in itemsDf.columns:
    cat_stats = itemsDf.groupby("Category")["Price"].agg(["count", "mean", "min", "max"])
    print("Category statistics:\n", cat_stats)

if "Brand" in itemsDf.columns:
    brand_stats = itemsDf.groupby("Brand")["Price"].mean()
    print("\nAverage price by brand:\n", brand_stats)
````

````python
# BANK SYSTEM — DO NOT MODIFY
# Read bank CSV into dictionary for in-memory manipulation
df = pd.read_csv("bank.csv")
bank = {}
for i in range(len(df)):
    account = str(df["AccountNumber"][i])
    bank[account] = {
        "FName": df["FName"][i],
        "LName": df["LName"][i],
        "Password": str(df["Password"][i]),
        "Address": df["Address"][i],
        "Class": df["Class"][i],
        "City": df["City"][i],
        "Balance": float(df["Balance"][i])
    }
while True:
    print("\n1. Create Account")
    print("2. Login")
    print("3. Exit")
    choice = input("Enter Choice: ")
    if choice == "1":
        account = input("Account Number: ")
        if account in bank:
            print("Account Already Exists")
        else:
            bank[account] = {}
            bank[account]["FName"] = input("First Name: ")
            bank[account]["LName"] = input("Last Name: ")
            bank[account]["Password"] = input("Password: ")
            bank[account]["Address"] = input("Address: ")
            bank[account]["Class"] = input("Class: ")
            bank[account]["City"] = input("City: ")
            bank[account]["Balance"] = float(input("Opening Balance: "))
            print("Account Created")

    elif choice == "2":
        account = input("Account Number: ")
        password = input("Password: ")
        if account in bank and password == bank[account]["Password"]:
            while True:
                print("\n1. View Details")
                print("2. Check Balance")
                print("3. Deposit")
                print("4. Withdraw")
                print("5. Logout")
                option = input("Choice: ")
                if option == "1":
                    print("First Name :", bank[account]["FName"])
                    print("Last Name :", bank[account]["LName"])
                    print("Address :", bank[account]["Address"])
                    print("Class :", bank[account]["Class"])
                    print("City :", bank[account]["City"])
                    print("Balance :", bank[account]["Balance"])
                elif option == "2":
                    print("Balance :", bank[account]["Balance"])
                elif option == "3":
                    amount = float(input("Amount: "))
                    bank[account]["Balance"] = bank[account]["Balance"] + amount
                    print("New Balance :", bank[account]["Balance"])
                elif option == "4":
                    amount = float(input("Amount: "))
                    if amount <= bank[account]["Balance"]:
                        bank[account]["Balance"] = bank[account]["Balance"] - amount
                        print("Remaining Balance :", bank[account]["Balance"])
                    else:
                        print("Insufficient Balance")
                elif option == "5":
                    break
                else:
                    print("Invalid Choice")
        else:
            print("Invalid Account or Password")
    elif choice == "3":
        accounts = []
        fname = []
        lname = []
        password = []
        address = []
        student_class = []
        city = []
        balance = []
        for account in bank:
            accounts.append(account)
            fname.append(bank[account]["FName"])
            lname.append(bank[account]["LName"])
            password.append(bank[account]["Password"])
            address.append(bank[account]["Address"])
            student_class.append(bank[account]["Class"])
            city.append(bank[account]["City"])
            balance.append(bank[account]["Balance"])
        df = pd.DataFrame({
            "AccountNumber": accounts,
            "FName": fname,
            "LName": lname,
            "Password": password,
            "Address": address,
            "Class": student_class,
            "City": city,
            "Balance": balance
        })
        df.to_csv("bank.csv", index=False)
        print("Data Saved")
        break
    else:
        print("Invalid Choice")
````

````python
# Using .apply() to transform data
# Categorize students by age group
def age_group(age):
    if age < 13:
        return "Child"
    elif age < 20:
        return "Teen"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

data["Age Group"] = data["age"].apply(age_group)
print("Data with age groups:\n", data)

print("\nAge group distribution:\n", data["Age Group"].value_counts())
````

## Using .apply() and .map()

- `.apply(func)` — apply a function to each element (or row/column)
- `.map(dict)` — replace values using a dictionary mapping

```python

df["age"].apply(lambda x: "Adult" if x >= 18 else "Minor")

# Map values
df["gender"].map({"M": "Male", "F": "Female"})
```
